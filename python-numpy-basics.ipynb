```python
arr1 = np.array([45,67,83,56,78,90,45,36])
print(f'First Four: {arr1[:4]}')
print(f'Middle: {arr1[2:6]}')
print(f'Every Second Value: {arr1[::2]}')
print(f'Reverse:{arr1[::-1]}')
```
