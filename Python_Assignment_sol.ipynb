# Q3.1

== checks whether two objects have the same value/content. is checks whether two variables point to the exact same object in memory (same id()). Two different objects can have equal values but different identities.

```python
# Q3.2

a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)   # True  -> same content
print(a is b)   # False -> two separate list objects in memory
```

```python
# Q3.3

x = 100
y = 100
print(x is y)   # True
print(x == y)   # True

p = 1000
q = 1000
print(p is q)   # False (typically)
print(p == q)   # True

## Python caches small integers (usually -5 to 256) as reusable singleton objects for performance, since they're used so often internally. So x and y (both 100) end up pointing to the same cached object, making is return True. Numbers outside that cached range, like 1000, are created as fresh objects each time, so p is q is usually False even though the values are equal. This is a CPython implementation detail, not a language guarantee — never rely on is to compare values, always use ==.
```

# Q5.5

The condition i < 10 is checked 11 times: it's True for i = 0 through i = 9 (10 checks that pass and run the body), and then checked once more when i = 10, where it evaluates False and the loop exits. 10 passing checks + 1 failing check = 11 total evaluations.

```python
# 6.1

for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
# Q6.2

for i in range(1, 16):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# Q6.3

for i in range(0, 5):
    if i % 3 == 0:
        print(i)
    elif i % 3 == 1:
        print(i + 10)
    else:
        print(i - 10)

## Trace: i=0 → 0%3=0 → prints 0. i=1 → 1%3=1 → prints 11. i=2 → 2%3=2 → else → prints -8. i=3 → 3%3=0 → prints 3. i=4 → 4%3=1 → prints 14. Output: 0, 11, -8, 3, 14
```

```python
# Q7.3

def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)

## print_abs prints the value but has no return statement, so it implicitly returns None. That None is what gets assigned to x.
```

```python
# Q7.4

def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)

## Output: 7. When inc_val(x) is called, the parameter val starts out bound to the same integer object as x (7). But integers are immutable — val = val + 1 doesn't modify that object, it creates a brand-new int object (8) and rebinds the local name val to it. x outside the function was never touched; it still points to the original 7.
```

```python
#Q8.2

counter = 0

def show_counter():
    print(counter)

show_counter()   # 0
```

```python
# Q8.3

def increment_counter():
    counter = counter + 1

increment_counter()

## This raises UnboundLocalError: local variable 'counter' referenced before assignment. Because the function body contains an assignment to counter (counter = ...), Python treats counter as a local variable for the entire function — even on the line where it's read, before the assignment happens. It never looks up the global counter at all.
```

```python
# Q8.4

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)   # 1

## The global keyword tells Python explicitly: "counter inside this function refers to the global variable, not a new local one" — so reads and writes both go to the global object.
```
