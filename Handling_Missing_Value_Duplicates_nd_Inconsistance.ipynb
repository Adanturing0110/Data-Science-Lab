# **Simple Statical Imputation**

**Mean Imputation**

```python
from sklearn.impute import SimpleImputer
mean_Imputer =SimpleImputer(strategy="mean")
df_mean = students.copy()
df_mean[["Attendance", "Marks"]] = mean_Imputer.fit_transform(df_mean[["Attendance", "Marks"]])
print(df_mean)
```

**Median Imputation**

```python
from sklearn.impute import SimpleImputer
median_Imputer =SimpleImputer(strategy="median")
df_median = students.copy()
df_median[["Attendance", "Marks"]] = mean_Imputer.fit_transform(df_median[["Attendance", "Marks"]])
print(df_median)
```

**Most Frequent-Value Imputation**

```python
from sklearn.impute import SimpleImputer

mode_Imputer = SimpleImputer(missing_values=None, strategy="most_frequent")
df_mode = students.copy()
df_mode[["Name"]] = mode_Imputer.fit_transform(df_mode[["Name"]])
print(df_mode)
```

```python
from sklearn.impute import SimpleImputer
const_Imputer =SimpleImputer(strategy="constant",fill_value=0)
df_const = students.copy()
df_const[["Attendance", "Marks"]] = const_Imputer.fit_transform(df_const[["Attendance", "Marks"]])
print(df_const)
```

# **KNN Imputation**

```python
from sklearn.impute import KNNImputer

knn_Imputer = KNNImputer(n_neighbors=2)
df_knn = students.copy()
df_knn[["Marks", "Attendance"]] = knn_Imputer.fit_transform(df_knn[["Marks", "Attendance"]])
print(df_knn)
```

# **Decide how to handle missing value**

```python
clean = students.copy()

clean["Marks"] = clean["Marks"].fillna(clean["Marks"].median())
clean["Attendance"]  = clean["Marks"].fillna(clean["Attendance"].median())
clean["Name"] = clean["Name"].fillna("Unknown")
print(clean)
```
