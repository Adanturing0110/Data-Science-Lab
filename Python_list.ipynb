# List
## List is an ordered collection of elements stored in a []
### mutable
### ordered
### allows the duplicates

```python
#create a list
marks = [98,87,56,76,77]
print(marks)
```

```python
student_info=[101,"Thor","God of Thunder",9.9]
print(student_info)
```

```python
#access the element using index
name = ["Anna","Ram","Krishna",'Sakura']
print(name)
print(name[3])
```

```python
#example for mutable in list
cgpa=[8.9,7.8,9.0,9.4]
print(cgpa)
cgpa[2]=8.3
print("after Updation:",cgpa)
```

```python
#add an item
Name=['Kiran',"Lalit","Humza","Shagun"]
Name.append("Deepika")
print(Name)
```

```python
#insert in particular index
Student_name=['Kiran',"Lalit","Humza","Shagun"]
Student_name.insert(1,"Asta")
print(Student_name)
```

```python
#extend function
sec_a=['A','B','C']
sec_b=['D','E','F']
sec_a.extend(sec_b)
print(sec_a)
print(type(sec_b))
```

```python
#remove an item in the list
sec_a_students=['Ashi','Binod','Charan','Dhanda','Ethanol','Fukra']
sec_a_students.remove('Charan')
print(sec_a_students)
del sec_a_students[0]
print(sec_a_students)
sec_a_students.pop()
print(sec_a_students)
```

```python
#list slicing
subjects = ["Python","DSA","OS","Computer Networks","SQL"]
print(subjects[0:3])
print(subjects[:4])
print(subjects[2:])

#length of the list
print(len(subjects))
```

```python
marks1=[78,67,70,45,39,51,89]
marks1.sort()
print(marks1)
marks1.sort(reverse=True)
print(marks1)
```

```python
#nested list
marks=[
    [91,67,87],
    [78,88,74],
    [56,78,69]
]
print(marks)
print(marks[1])
print(marks[1][2])
```
